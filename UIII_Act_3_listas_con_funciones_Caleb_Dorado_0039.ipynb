## Ejemplos de listas con funciones

```python
# ==========================================
# SISTEMA DE PANADEROS EN PYTHON
# ==========================================

panaderos = []

# ==========================================
# FUNCIÓN AGREGAR
# ==========================================
def agregar_panadero():
    print("\n--- REGISTRAR NUEVO PANADERO ---")
    nombre = input("Nombre del panadero/maestro: ")
    especialidad = input("Especialidad (ej. Masa madre, Bollería): ")
    pan_estrella = input("Pan estrella: ")
    experiencia = input("Años de experiencia: ")

    panadero = {
        "nombre": nombre,
        "especialidad": especialidad,
        "pan_estrella": pan_estrella,
        "experiencia": experiencia
    }

    panaderos.append(panadero)
    print(f"¡El maestro {nombre} ha sido registrado en el gremio!")

# ==========================================
# FUNCIÓN LEER (POR NOMBRE)
# ==========================================
def leer_panadero():
    print("\n--- CONSULTAR FICHA DEL PANADERO ---")
    if not panaderos:
        print("El obrador está vacío (no hay panaderos).")
        return

    buscar = input("Nombre del panadero que busca: ")
    encontrado = False

    for p in panaderos:
        if p["nombre"].lower() == buscar.lower():
            print("\n--- DATOS DE LA TAHONA ---")
            print(f"Nombre: {p['nombre']}")
            print(f"Especialidad: {p['especialidad']}")
            print(f"Pan Estrella: {p['pan_estrella']}")
            print(f"Experiencia: {p['experiencia']} años")
            encontrado = True
            break

    if not encontrado:
        print(f"No se encontró a ningún panadero llamado '{buscar}'.")

# ==========================================
# FUNCIÓN LISTAR
# ==========================================
def listar_panaderos():
    print("\n--- NÓMINA DE PANADEROS ---")
    if not panaderos:
        print("No hay panaderos amasando todavía.")
    else:
        for p in panaderos:
            print(f" >> {p['nombre']} - Especialista en {p['especialidad']} (Estrella: {p['pan_estrella']})")

# ==========================================
# FUNCIÓN MODIFICAR
# ==========================================
def modificar_panadero():
    print("\n--- ACTUALIZAR TÉCNICAS ---")
    buscar = input("Nombre del panadero a editar: ")

    for p in panaderos:
        if p["nombre"].lower() == buscar.lower():
            print(f"Editando la ficha de {p['nombre']}. Deje en blanco para no cambiar.")

            nuevo_nombre = input(f"Nuevo nombre [{p['nombre']}]: ")
            nueva_esp = input(f"Nueva especialidad [{p['especialidad']}]: ")
            nuevo_pan = input(f"Nuevo pan estrella [{p['pan_estrella']}]: ")
            nueva_exp = input(f"Nuevos años de experiencia [{p['experiencia']}]: ")

            if nuevo_nombre: p["nombre"] = nuevo_nombre
            if nueva_esp: p["especialidad"] = nueva_esp
            if nuevo_pan: p["pan_estrella"] = nuevo_pan
            if nueva_exp: p["experiencia"] = nueva_exp

            print("¡Receta actualizada con éxito!")
            return

    print("Panadero no encontrado en el sistema.")

# ==========================================
# FUNCIÓN BORRAR
# ==========================================
def borrar_panadero():
    print("\n--- DAR DE BAJA PANADERO ---")
    buscar = input("Nombre del panadero a eliminar: ")

    for p in panaderos:
        if p["nombre"].lower() == buscar.lower():
            panaderos.remove(p)
            print(f"El maestro {buscar} ha dejado el obrador.")
            return

    print("No se pudo eliminar: El panadero no está en la lista.")

# ==========================================
# MENÚ PRINCIPAL
# ==========================================
while True:
    print("\n" + "="*35)
    print("      GESTIÓN DE PANADEROS v1.0")
    print("="*35)
    print("1. Agregar panadero")
    print("2. Consultar ficha (por nombre)")
    print("3. Ver lista de maestros")
    print("4. Modificar especialidad/datos")
    print("5. Eliminar del gremio")
    print("6. Cerrar panadería (Salir)")
    print("="*35)

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        agregar_panadero()
    elif opcion == "2":
        leer_panadero()
    elif opcion == "3":
        listar_panaderos()
    elif opcion == "4":
        modificar_panadero()
    elif opcion == "5":
        borrar_panadero()
    elif opcion == "6":
        print("Cerrando el horno... ¡Hasta pronto!")
        break
    else:
        print("Esa opción se ha quemado (no existe). Intenta de nuevo.")

print("Hecho por Caleb 0039")
```
