## Ejemplos de listas con funciones

```python
# --- BASE DE DATOS DE LA PANADERÍA ---
# Cada panadero se guardará como una sublista: [ID, Nombre, Especialidad, Experiencia]
panaderos = []

# --- FUNCIONES DE "PAN-IA" (SIMULADOR DE IA) ---

def simular_ia_analisis():
    """Simula una IA que analiza la situación actual de la panadería"""
    total = len(panaderos)
    print("\n[🧠 PanIA Analizando]:", end=" ")
    if total == 0:
        print("La panadería está vacía. Sugiero contratar a un experto en 'Pan de Muerto' o 'Conchas'.")
    elif total < 3:
        print(f"Tienes {total} panaderos. El equipo es pequeño, evita saturarlos con pedidos de hojaldre.")
    else:
        print(f"¡Equipo óptimo! Con {total} panaderos podemos expandir el menú a pasteles gourmet.")


# --- RECURSIÓN PARA REEMPLAZAR EL CICLO FOR ---

def mostrar_tabla_recursiva(lista, indice):
    """Sustituye al ciclo for para imprimir la tabla de panaderos"""
    if indice >= len(lista):
        return # Caso base: ya no hay más panaderos

    panadero = lista[indice]
    # Usamos formato de texto para alinear perfectamente las columnas
    print(f"| {panadero[0]:<5} | {panadero[1]:<15} | {panadero[2]:<20} | {panadero[3]:<10} |")
    mostrar_tabla_recursiva(lista, indice + 1)


# --- FUNCIONES PRINCIPALES ---

def agregar_panadero():
    print("\n--- 🥖 AGREGAR NUEVO PANADERO ---")
    id_panadero = input("ID único (ej. 101): ")
    nombre = input("Nombre del panadero: ")
    especialidad = input("Especialidad (ej. Conchas, Bolillos): ")
    experiencia = input("Años de experiencia: ")

    # Agregamos al panadero como una lista dentro de la lista principal
    panaderos.append([id_panadero, nombre, especialidad, experiencia])
    print(f"✨ ¡PanIA ha registrado con éxito a {nombre}!")
    simular_ia_analisis()


def leer_panadero(panadero):
    """Muestra la información individual de un panadero encontrado"""
    print("\n--- 📋 FICHA TÉCNICA DEL PANADERO ---")
    print(f"🆔 ID: {panadero[0]}")
    print(f"👤 Nombre: {panadero[1]}")
    print(f"🥐 Especialidad: {panadero[2]}")
    print(f"⭐ Experiencia: {panadero[3]} años")

    # Consejo de IA basado en la experiencia
    if int(panadero[3]) > 5:
        print("💡 [PanIA Tip]: Este panadero es Maestro. Asígnalo a recetas complejas.")
    else:
        print("💡 [PanIA Tip]: Panadero junior. Buen elemento para apoyo y horneado básico.")


def listar_panaderos():
    print("\n=================================================================")
    print("                 📜 PERSONAL DE LA PANADERÍA                     ")
    print("=================================================================")
    print(f"| {'ID':<5} | {'Nombre':<15} | {'Especialidad':<20} | {'Exp. (Años)':<10} |")
    print("-----------------------------------------------------------------")

    if len(panaderos) == 0:
        print(f"| {'NO HAY PANADEROS REGISTRADOS AÚN':^61} |")
    else:
        # Iniciamos la función recursiva desde el índice 0
        mostrar_tabla_recursiva(panaderos, 0)

    print("=================================================================")
    simular_ia_analisis()


def buscar_recursivo(lista, id_buscar, indice):
    """Busca un panadero por ID sin usar ciclos for"""
    if indice >= len(lista):
        return None # No se encontró
    if lista[indice][0] == id_buscar:
        return lista[indice] # Se encontró, regresa al panadero
    return buscar_recursivo(lista, id_buscar, indice + 1)


def buscar_panadero():
    print("\n--- 🔍 BUSCAR PANADERO ---")
    id_buscar = input("Introduce el ID del panadero que deseas buscar: ")
    panadero_encontrado = buscar_recursivo(panaderos, id_buscar, 0)

    if panadero_encontrado:
        leer_panadero(panadero_encontrado)
    else:
        print("❌ PanIA no encontró ningún panadero con ese ID.")


def modificar_panadero():
    print("\n--- 🛠️ MODIFICAR PANADERO ---")
    id_buscar = input("Introduce el ID del panadero a modificar: ")
    panadero = buscar_recursivo(panaderos, id_buscar, 0)

    if panadero:
        print(f"Modificando a: {panadero[1]}")
        panadero[1] = input("Nuevo Nombre: ")
        panadero[2] = input("Nueva Especialidad: ")
        panadero[3] = input("Nuevos Años de Experiencia: ")
        print("✅ ¡Datos actualizados correctamente por PanIA!")
    else:
        print("❌ No se encontró el panadero para modificar.")


def borrar_panadero():
    print("\n--- 🗑️ DESPEDIR / BORRAR PANADERO ---")
    id_buscar = input("Introduce el ID del panadero a eliminar: ")
    panadero = buscar_recursivo(panaderos, id_buscar, 0)

    if panadero:
        panaderos.remove(panadero)
        print(f"🔥 El panadero ha sido removido del sistema.")
        simular_ia_analisis()
    else:
        print("❌ El ID no existe. Nadie fue eliminado.")


# --- MENÚ INTERACTIVO RECURSIVO ---

def menu():
    print("\n=== 🤖 SISTEMA INTELIGENTE 'PanIA' v1.0 ===")
    print("1. Agregar Panadero 🥖")
    print("2. Listar Panaderos 📜")
    print("3. Buscar Panadero 🔍 (Ficha Técnica)")
    print("4. Modificar Panadero 🛠️")
    print("5. Borrar Panadero 🗑️")
    print("6. Salir del Sistema 🚪")

    opcion = input("Selecciona una opción (1-6): ")

    if opcion == "1":
        agregar_panadero()
        menu() # Volver a llamar al menú sin usar ciclos
    elif opcion == "2":
        listar_panaderos()
        menu()
    elif opcion == "3":
        buscar_panadero()
        menu()
    elif opcion == "4":
        modificar_panadero()
        menu()
    elif opcion == "5":
        borrar_panadero()
        menu()
    elif opcion == "6":
        print("\n¡Gracias por usar PanIA! Que tengas un dulce día. 🥐🤖")
    else:
        print("⚠️ Opción no válida, intenta de nuevo.")
        menu()

# --- ARRANCAR EL PROGRAMA ---
# Al inicio la lista 'panaderos' está vacía
menu()
```
